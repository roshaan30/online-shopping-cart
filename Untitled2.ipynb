```python
# Online Shopping Cart System
# List
products = ["Hair Dryer","Television","Dining Table","Vacuum Cleaner","Mirror"]
print("Products:", products)

# tuple  
prices = (("Hair Dryer",1500),("Television",39999),("Dining Table",12000),
          ("Vacuum Cleaner",4000),("Mirror",2000))
# set
categories = {"Electronics","Furnitures","Home appliances"}
print ("Categories:",categories)
# dict
cart = {}
while True:
    choice = input("Enter product name or exit: ").title()
    if choice == "Exit":
        break
    if choice not in products:
        print("Product unavailable!")
        continue
    qty = int(input("Enter quantity: "))
    if qty <= 0:
        print("Invalid quantity!")
        continue
    cart[choice] = cart.get(choice, 0) + qty
total = 0
for product, qty in cart.items():
    for name, price in prices:
        if product == name:
            amount = price * qty
            total += amount
            print(product,"x", qty,"=","Rs.",amount)
print("--------------------------")
print("Total bill: Rs.", total)
if total >= 5000:
    total *= 0.9
    print("10% discount applied!")
print("Final bill: Rs.", total)
print("Thank you for shopping!")
```
